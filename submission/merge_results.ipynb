```python

from constrerl.erl_schema import Annotated, AnnotatedArticle
import glob
from pathlib import Path
import json
import pandas as pd
from collections.abc import Callable, Awaitable
```

```python
set_type = "test"
results_dir = Path(f"../data/results_{set_type}")
merge_dir = Path(f"../data/merge_{set_type}")

onto_file_dir = Path("../data/results_onto/Graphwise_GutBrainIE_2026")
```

```python
def load_results(result_file: Path) -> dict[str, AnnotatedArticle]:
    with open(result_file) as f:
        data: dict[str, dict] = json.load(f)
        return {
            article_id: AnnotatedArticle.model_validate(article_dict)
            for article_id, article_dict in data.items()
        }
```

```python
onto_results = {
    "T611": ["Graphwise_T611_24_ENS-EXP-4-SUB-1", "Graphwise_T611_13_NER-EXP-8-SUB-1"],
    "T612": [
        "Graphwise_T612_2_24-ENS-EXP-4-SUB-1-EL-freq-gaze",
        "Graphwise_T612_3_13-NER-EXP-8-SUB-1-EL-freq-gaze",
    ],
    "T621": ["Graphwise_T621_BGPT5520G_gpt-5.5_20_shot_gold"],
    "T622": ["Graphwise_T622_BGPT5520G_gpt-5.5_20_shot_gold"],
}
tug_results = {
    "T611": [
        "eval_naive-entities",
        "eval_hermes_neefinetuned-lora-entities-naive-beam-none-hermes-3-2-3B-entities",
    ],
    "T612": [
        "eval_naive-entities",
        "eval_hermes_neefinetuned-lora-entities-naive-beam-none-hermes-3-2-3B-entities",
    ],
    "T621": ["eval_hermes-lora-relations-beam-none-hermes-3-2-3B-relations"],
    "T622": ["eval_hermes-lora-relations-beam-none-hermes-3-2-3B-relations"],
}


def load_onto_results(f: str) -> dict[str, AnnotatedArticle]:
    result_files = list((onto_file_dir / f).glob("*.json"))
    results = load_results(result_files[0]) if len(result_files) > 0 else {}
    return results


def load_tug_results(f: str) -> dict[str, AnnotatedArticle]:
    result_file = results_dir / f"{f}.json"
    results = load_results(result_file) if result_file.exists() else {}
    return results


onto_results_loaded = {
    t: [(f,load_onto_results(f)) for f in fs] for t, fs in onto_results.items()
}
tug_results_loaded = {
    t: [(f,load_tug_results(f)) for f in fs] for t, fs in tug_results.items()
}
```

```python
task_fields ={
    "T611": ["start_idx", "end_idx", "text_span", "location", "label"],
    "T612": ["start_idx", "end_idx", "text_span", "location", "label", "uri"],
    "T621": ["subject_text_span", "subject_label", "predicate", "object_text_span", "object_label"],
    "T622": ["subject_uri", "subject_label", "predicate", "object_uri", "object_label"],
}
taks_elem = {
    "T611": "entities",
    "T612": "entities",
    "T621": "mention_level_relations",
    "T622": "concept_level_relations",
}
```

```python
merge_dir.mkdir(exist_ok=True)
def merge_results(
    onto_result: dict[str, AnnotatedArticle],
    tug_result: dict[str, AnnotatedArticle],
    k: str,
) -> dict[str, dict[str, AnnotatedArticle]]:
    if set(onto_result.keys()) != set(tug_result.keys()):
        print(
            f"Warning: Article IDs do not match for {k} between Onto and TUG results."
        )
        return {}
    combined_results = {}
    relevant_fields = task_fields[k]

    for art_id, onto_article in onto_result.items():
        tug_article = tug_result.get(art_id)
        if tug_article is None:
            print(
                f"Warning: Article ID {art_id} found in Onto results but not in TUG results for {k}."
            )
            continue
        onto_article_dict = onto_article.model_dump()
        if onto_article_dict[taks_elem[k]] is None:
            onto_article_dict[taks_elem[k]] = []
        tug_article_dict = tug_article.model_dump()
        if tug_article_dict[taks_elem[k]] is None:
            tug_article_dict[taks_elem[k]] = []
        ie_onto = [
            tuple([i[key] for key in relevant_fields])
            for i in onto_article_dict[taks_elem[k]]
        ]
        ie_tug = [
            tuple([i[key] for key in relevant_fields])
            for i in tug_article_dict[taks_elem[k]]
        ]
        ie_sets = {
            "union": set(ie_tug).union(ie_onto),
            "intersection": set(ie_tug).intersection(ie_onto),
        }
        for n, ie_set in ie_sets.items():
            if combined_results.get(n) is None:
                combined_results[n] = {}
            print(f"Set {n}: {len(ie_set)} unique relations")
            # combine results for this article and set
            combined_ie = []
            for id_tuple in ie_set:
                combined_ie.append(
                    {field: id_tuple[idx] for idx, field in enumerate(relevant_fields)}
                )

            combined_results[n][art_id] = {taks_elem[k]: combined_ie}
    return combined_results

for k, onto_result_list in onto_results_loaded.items():
    for k, tug_result_list in tug_results_loaded.items():
        if k != k:
            continue
        for f_onto, onto_result in onto_result_list:
            for f_tug, tug_result in tug_result_list:
                combined_results = merge_results(onto_result, tug_result, k)
                if combined_results is None:
                    print(f"Skipping {k}/{f_onto}/{f_tug} due to mismatched article IDs.")
                    continue
                for set_type, combined_result in combined_results.items():
                    with open(
                        merge_dir / f"{f_tug}_{f_onto}_{set_type}.json", "w"
                    ) as f:
                        json.dump(combined_result, f, indent=4)
```
